```python
import ast
genres=[]
def convert(s):
    l1=[]
    for i in ast.literal_eval(s):
        l1.append(i['name'])
        if i['name'] not in genres:
            genres.append(i['name'])
    text=l1[:]
   
    l1=[]
    return text
```

```python
def fetch_cast(text):
    L=[]
    counter=0
    for i in ast.literal_eval(text):
            if counter<3:
                L.append(i['name'])
            else:
                break
            counter+=1
    L1=L[:]
    L.clear()
    return L1
```

```python
final['cast']=final['cast'].apply(fetch_cast)
```

```python
final.head()
```

```python
def fetch_director(text):
    L=[]
    for i in ast.literal_eval(text):
        if i['job']=='Director':
            L.append(i['name'])
            break
    L1=L[:]
    L.clear()
    return L1
```

```python
final['crew']=final['crew'].apply(fetch_director)
```

```python
final.head()
```

```python
final['tags']=final['genres']+final['keywords']+final['overview']+final['cast']+final['crew']
```

```python
def remove_space(L):
    L1=[]
    for i in L:
        L1.append(i.replace(' ',''))
    
    L2=L1[:] 
    L1.clear()
    return L2
```

```python
final['tags']=final['tags'].apply(remove_space)
```

```python
final.head()
```

```python
final['tags']=final['tags'].apply(lambda x:" ".join(x))
```

```python
#final recommendation function for both movie name and actor name
def recommend(movie_name):
    sim = []
    recs=[]
    flag=0
    movie_name=movie_name.lower()
    
    #checking if movie name is present
    if movie_name in mov:
        flag=1
        
        movie_index=final[final['lower_title']== movie_name].index[0]
        
    #case when actor name is entered, checking if its present
    else:
        actor_name=movie_name.lower()
        for i in range(len(listt_cast)):
            if actor_name in listt_cast[i]:
                
                flag=2
                
                title= final.iloc[i]['lower_title']
                break
        if flag==2:
            movie_index=final[final['lower_title']== title].index[0]
        
        
    if flag==1 or flag==2:
        
        #finding movies similar to given movie
        L=sorted(list(enumerate(similarity_matrix[movie_index])),reverse=True,key=lambda x:x[1])
        #print(type(L))
        l = np.array(L)
        
        
        #recommending top 5 most similar movies
        for i in L[1:6]:
            print(final.iloc[i[0]]['original_title'])
            recs.append(final.iloc[i[0]]['original_title'])
        
        for i in range(1, 6):
            sim.append(l[i,1])
            
        print(sim)
        
        # Average of cosine similarities
        print(np.mean(sim))
        return recs
    
    #if movie/actor name not present in dataset
    else:
        return recs
        
```

```python
input_actors = ['daniel craig','chris evans','will smith','johnny depp','meryl streep','scarlett johansson','emma watson','emma stone','jennifer lawrence','zac efron']
expected_movies = ['Skyfall','Captain America: Civil War','Men in Black','Pirates of the Caribbean: The Curse of the Black Pearl','The Devil Wears Prada','Avengers: Age of Ultron','Harry Potter and the Chamber of Secrets','The Amazing Spider-Man 2','X-Men','Baywatch']
accuracy = []
all_predictions = []
for i in range(len(input_actors)):
    predictions = recommend(input_actors[i])
    all_predictions.append(predictions)
    accuracy.append(expected_movies[i] in predictions)
```
