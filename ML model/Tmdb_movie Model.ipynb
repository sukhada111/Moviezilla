```python
import numpy as np
import pandas as pd
```

```python
movie=pd.read_csv('tmdb_5000_movies.csv')
```

```python
credits=pd.read_csv('tmdb_5000_credits.csv')
```

```python
#merging datasets
final=pd.merge(movie,credits,left_on='id',right_on='movie_id')[['id','original_title','genres','keywords','overview','cast','crew']]
```

```python
final.head()
```

```python
final.isnull().sum()
```

```python
final.dropna(subset=['overview'],inplace=True)
```

```python
final.isnull().sum()
```

```python
final['genres'][0]
```

```python
import ast
genres=[]
def convert(s):
    l1=[]
    for i in ast.literal_eval(s):
        l1.append(i['name'])
        if i['name'] not in genres:
            genres.append(i['name'])
    text=l1[:]
   
    l1=[]
    return text
```

```python
final['genres']=final['genres'].apply(convert)
```

```python
final['genres'].head()
```

```python
print(genres)
```

```python
print(len(genres))
```

```python
final['keywords']=final['keywords'].apply(convert)
```

```python
final['keywords'].head()
```

```python
final['overview']=final['overview'].apply(lambda x : x.split(' '))
```

```python
final.head()
```

```python
def fetch_cast(text):
    L=[]
    counter=0
    for i in ast.literal_eval(text):
            if counter<3:
                L.append(i['name'])
            else:
                break
            counter+=1
    L1=L[:]
    L.clear()
    return L1
```

```python
final['cast']=final['cast'].apply(fetch_cast)
```

```python
final.head()
```

```python
def fetch_director(text):
    L=[]
    for i in ast.literal_eval(text):
        if i['job']=='Director':
            L.append(i['name'])
            break
    L1=L[:]
    L.clear()
    return L1
```

```python
final['crew']=final['crew'].apply(fetch_director)
```

```python
final.head()
```

```python
final['tags']=final['genres']+final['keywords']+final['overview']+final['cast']+final['crew']
```

```python
def remove_space(L):
    L1=[]
    for i in L:
        L1.append(i.replace(' ',''))
    
    L2=L1[:] 
    L1.clear()
    return L2
```

```python
final['tags']=final['tags'].apply(remove_space)
```

```python
final.head()
```

```python
final['tags']=final['tags'].apply(lambda x:" ".join(x))
```

```python
final.head()
```

```python
final['tags']=final['tags'].apply(lambda x:x.lower())
```

```python
final.head()
```

```python
def convert(name):
    return name.lower()

final['lower_title']=final['original_title'].apply(convert)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv=CountVectorizer(stop_words='english',max_features=10000)
```

```python
X = cv.fit_transform(final['tags']).toarray()
```

```python
X.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_matrix = cosine_similarity(X)
```

```python
import pickle
pickle.dump(cv, open('cv.pkl','wb'))
pickle.dump(similarity_matrix, open('similarity_matrix.pkl','wb'))
```

```python
similarity_matrix.shape
```

```python
# finding the index of the movie
final[final['original_title']=='Avengers: Age of Ultron'].index[0]
```

```python
mov=np.array(final['lower_title'])
```

```python
def recommend(movie_name):
    recs=[]
    movie_name=movie_name.lower()
    if movie_name in mov:
        movie_index=final[final['lower_title']== movie_name].index[0]

        L=sorted(list(enumerate(similarity_matrix[movie_index])),reverse=True,key=lambda x:x[1])

        
        for i in L[1:6]:
            print(final.iloc[i[0]]['original_title'])
            recs.append(final.iloc[i[0]]['original_title'])
        return recs
    #if movie name not present in dataset
    else:
        return recs
```
